# Why Alakazam is a Good Rule-Based Baseline for PTCG AI Battle

After reaching the top 3 with Alakazam, I wanted to write down a few thoughts on my deck choice and why this archetype feels like a strong rule-based baseline for the Pokémon TCG AI Battle competition.

## Why Alakazam?

For a rule-based agent, I think Alakazam is one of the most natural decks to start with.

The deck has a clear and repeatable game plan: set up the Alakazam evolution line, attach Psychic Energy, maintain a follow-up attacker, and convert hand size into damage.

This makes the deck easier to reason about than many more flexible or combo-heavy strategies. The agent does not need to understand every possible opponent plan at first. It mainly needs to make its own turns coherent.

## Basic Game Plan

The basic plan is simple:

- Build Abra into Kadabra and then Alakazam.
- Keep enough Psychic Energy available to attack every turn.
- Prepare the next Alakazam before the current attacker is Knocked Out.
- Use draw and search cards to keep the hand large enough for damage.
- Avoid spending recovery cards too early or too late.

In practice, the most important point is continuity. Alakazam needs to attack repeatedly, so the deck must avoid running out of attackers in the late game.

## Why This Fits a Rule-Based Agent

Alakazam has many decisions that can be expressed as priorities:

- Can I attack this turn?
- Can I make an Alakazam this turn?
- Do I have a next attacker ready?
- Do I need Energy now or later?
- Is drawing more cards useful, or dangerous?
- Can I take a meaningful Prize this turn?

These questions are simple, but they cover a large part of what the deck needs to do.

For this reason, I think Alakazam is a good example of an important lesson in this competition: before trying to outplay every opponent, the agent should first learn to play its own deck correctly.

## Why It Works Well Early

Alakazam is not an unbeatable deck. It has clear weaknesses, and there are decks that can punish it.

However, in the early stage of this competition, many of those counter-strategies are harder to implement as reliable agents. They often require more complex sequencing, more matchup-specific decisions, or more precise resource management.

By contrast, Alakazam has a relatively stable baseline plan. If the agent can execute its own setup, maintain the attacker chain, and manage resources reasonably well, it can perform consistently even without highly sophisticated opponent modeling.

This is one reason I think Alakazam is a strong early competition choice: not because it has no counters, but because its own game plan is easier to make reliable.

## What Is Hard?

Even with a clear deck like Alakazam, there are still many non-trivial decisions.

One important rule of thumb is that attacking should usually be the final action of the turn. Before attacking, the agent should check whether it can evolve, attach Energy, prepare the Bench, recover resources, or improve the Prize route.

Another challenge is resource management. Drawing cards is usually good, but not always. In longer games, drawing too aggressively can create deck-out risk or leave the agent without enough resources for the final attacks.

Recovery cards are also difficult. Using them too early may waste value, but waiting too long can leave the deck without a follow-up attacker.

## Key Lesson

For me, Alakazam shows that a strong baseline agent does not have to begin with very complex opponent prediction.

A good first goal is more fundamental:

> Make the deck execute its own plan correctly.

Once that baseline is solid, matchup-specific improvements become much easier to add.
